**Round Robin**

```python

n=int(input("Enter number of process: "))
burst_times=[]

for i in range (n):
  bt=int(input("Enter Bust Time: "))
  burst_times.append(bt)

quantum=int(input("Enter the quantum time:"))
rem_bt=list(burst_times)
wt=[0]*n
current_time=0

while True:
  all_done=True

  for i in range(n):
    if rem_bt [i] > 0:
      all_done=False

      if rem_bt[i] > quantum :
        current_time += quantum
        rem_bt[i] -= quantum

      else:
          current_time +=rem_bt[i]
          wt[i] = current_time - burst_times[i]
          rem_bt[i] = 0

  if all_done:
    break

print("Process | Burst Time | Wait Time | Turnaround Time")

total_tat=0

for i in range (n):
  Turnaround= burst_times[i]+ wt[i]
  total_tat += Turnaround
  print(f"  p{i+1}   |   {burst_times[i]}   |   {wt[i]}   |   {Turnaround}")

avg_tat=total_tat/n

print("Total Turnaround Time =", total_tat)
print("Average Turnaround Time =", avg_tat)



```
