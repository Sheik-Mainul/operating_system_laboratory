**Page Algorithm**

```python
frames = []
faults = 0

n = int(input("Frames: "))
pages = list(map(int, input("Pages: ").split()))

for p in pages:
    if p not in frames:
        if len(frames) == n:
            frames.pop(0)
        frames.append(p)
        faults += 1
    print(f"Page {p} -> {frames}")

print("Total Page Faults:", faults)
```

**Most Recent Use**

```python
frames = []
faults = 0

n = int(input("Frames: "))
pages = list(map(int, input("Pages: ").split()))

for p in pages:
    if p not in frames:
        if len(frames) == n:
            frames.pop()
        frames.append(p)
        faults += 1
    else:
        frames.remove(p)
        frames.append(p)

    print(f"Page {p} : {frames}")

print("Total Page Faults:", faults)
```
