**First Come First Serve (FCFS)**

FCFS.jpg

```python



at = []
bt = []

for i in range(4):
    a = int(input("Enter AT: "))
    b = int(input("Enter BT: "))
    at.append(a)
    bt.append(b)

current_time = 0
total_tat = 0
total_wt = 0

print("\nAT\tBT\tCT\tTAT\tWT")

for i in range(4):

    if current_time < at[i]:
        current_time = at[i]

    ct = current_time + bt[i]
    tat = ct - at[i]
    wt = tat - bt[i]

    current_time = ct
    total_tat = total_tat + tat
    total_wt = total_wt + wt

    print(at[i], "\t", bt[i], "\t", ct, "\t", tat, "\t", wt)

print("\nAVG TAT =", total_tat / 4)
print("AVG WT =", total_wt / 4)




```
