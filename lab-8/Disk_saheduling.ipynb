 **FCFS** (First Come First Served)

```python
req_seq = [176,51,50,40,49,100]
head=35
total_seek = 0
for i in req_seq :
  total_seek += abs(i - head)
  head=i
  print("Total head moment:",total_seek)
```

**SSTF** (Shortest Seek Time First)

```python
head= int(input("Enter initial head position: "))
req_seq= list(map(int, input("Enter request sequence (Use space): ").split()))

total_seek= 0
current= head
requests= req_seq.copy()
seek_sequence= [current]

while requests:
    closest= min(requests, key=lambda x: abs(x - current))

    total_seek += abs(closest - current)
    current = closest
    seek_sequence.append(current)
    requests.remove(closest)

print("\nSeek Sequence:", " -> ".join(map(str, seek_sequence)))
print("Total head movement:", total_seek)
```

**SCAN**

```python
head = int(input("Head position: "))
reqs = list(map(int, input("Requests: ").split()))
direction = input("Direction (left or right): ").strip().lower()

left = sorted([x for x in reqs if x < head], reverse=True)
right = sorted([x for x in reqs if x >= head])

if direction == "left":
    if 0 not in left:
        left.append(0)
    path = left + right
else:
    if 99 not in right:
        right.append(99)
    path = right + left

total_movement = 0
current = head

for track in path:
    total_movement += abs(track - current)
    current = track

print("Order:", [head] + path)
print("Total head movement:", total_movement)
```
