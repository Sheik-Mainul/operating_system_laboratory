**Priority Scheduling (Non-Preemptive )**

```python

n = int(input("Enter the number of processes: "))

p_id = []
at = []
bt = []
pr = []
completed = []
ct = [0] * n
wt = [0] * n
tat = [0] * n

for i in range(n):
    p_id.append(i + 1)
    at.append(int(input("Enter Arrival Time: ")))
    bt.append(int(input("Enter Burst Time: ")))
    pr.append(int(input("Enter Priority (Lower number = Higher priority): ")))
    completed.append(False)

current_time = 0
completed_count = 0

print("\nExecution Order: ", end="")

while completed_count < n:
    highest_priority = 9999
    selected = -1

    for i in range(n):
        if completed[i] == False and at[i] <= current_time and pr[i] < highest_priority:
            highest_priority = pr[i]
            selected = i

    if selected != -1:
        print(f"P{p_id[selected]}", end=" ")
        current_time = current_time + bt[selected]
        ct[selected] = current_time
        tat[selected] = ct[selected] - at[selected]
        wt[selected] = tat[selected] - bt[selected]
        completed[selected] = True
        completed_count = completed_count + 1
    else:
        current_time = current_time + 1

print("\n")
print("Process\tAT\tBT\tCT\tTAT\tWT")
for i in range(n):
    print(f"P{p_id[i]}\t{at[i]}\t{bt[i]}\t{ct[i]}\t{tat[i]}\t{wt[i]}")

avg_wt = sum(wt) / n
avg_tat = sum(tat) / n

print(f"\nAverage Waiting Time = {avg_wt}")
print(f"Average Turnaround Time = {avg_tat}")

```

**Priority Scheduling (Preemptive )**

```python
n = int(input("Enter the number of processes: "))

p_id = []
at = []
bt = []
rt = []
pr = []
ct = [0] * n
wt = [0] * n
tat = [0] * n

for i in range(n):

    p_id.append(i + 1)

    arrival = int(input("Enter Arrival Time: "))
    burst = int(input("Enter Burst Time: "))
    priority = int(input("Enter Priority (Lower number = Higher priority): "))

    at.append(arrival)
    bt.append(burst)
    rt.append(burst)
    pr.append(priority)

current_time = 0
completed_count = 0

print("\nExecution Timeline (per 1 unit of time): ", end="")

while completed_count < n:
    highest_priority = 9999
    selected = -1

    for i in range(n):
        if rt[i] > 0 and at[i] <= current_time and pr[i] < highest_priority:
            highest_priority = pr[i]
            selected = i

    if selected != -1:
        print(f"P{p_id[selected]}", end=" ")
        rt[selected] = rt[selected] - 1
        current_time = current_time + 1

        if rt[selected] == 0:
            ct[selected] = current_time
            tat[selected] = ct[selected] - at[selected]
            wt[selected] = tat[selected] - bt[selected]
            completed_count = completed_count + 1
    else:
        print("IDLE", end=" ")
        current_time = current_time + 1

print("\n")
print("Process\tAT\tBT\tCT\tTAT\tWT")
for i in range(n):
    print(f"P{p_id[i]}\t{at[i]}\t{bt[i]}\t{ct[i]}\t{tat[i]}\t{wt[i]}")

avg_wt = sum(wt) / n
avg_tat = sum(tat) / n

print(f"\nAverage Waiting Time = {avg_wt}")
print(f"Average Turnaround Time = {avg_tat}")
```
