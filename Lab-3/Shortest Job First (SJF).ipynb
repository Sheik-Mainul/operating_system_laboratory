**Shortest Job First (SJF) Non-Preemptive**
                                                          

```python


n = int(input("Enter number of processes: "))

at = []
bt = []

for i in range(n):
    a = int(input(f"Enter AT for P{i+1}: "))
    b = int(input(f"Enter BT for P{i+1}: "))
    at.append(a)
    bt.append(b)

ct = [0] * n
tat = [0] * n
wt = [0] * n
done = [False] * n

completed = 0
current_time = 0

while completed < n:
    idx = -1
    min_bt = 999999

    for i in range(n):
        if not done[i] and at[i] <= current_time:
            if bt[i] < min_bt:
                min_bt = bt[i]
                idx = i

    if idx != -1:
        current_time = current_time + bt[idx]
        ct[idx] = current_time
        tat[idx] = ct[idx] - at[idx]
        wt[idx] = tat[idx] - bt[idx]

        done[idx] = True
        completed = completed + 1
    else:
        current_time = current_time + 1


print("\nPID\tAT\tBT\tCT\tTAT\tWT")
for i in range(n):
    print(f"P{i+1}\t{at[i]}\t{bt[i]}\t{ct[i]}\t{tat[i]}\t{wt[i]}")

print("\nAVG TAT =", sum(tat) / n)
print("AVG WT =", sum(wt) / n)


```

**Shortest Job First (SJF) Preemptive**
                                                          

```python


n = int(input("Enter number of processes: "))

at = []
bt = []
rt = []

for i in range(n):
    a = int(input(f"Enter AT for P{i+1}: "))
    b = int(input(f"Enter BT for P{i+1}: "))
    at.append(a)
    bt.append(b)
    rt.append(b)

ct = [0] * n
tat = [0] * n
wt = [0] * n

completed = 0
current_time = 0

while completed < n:
    idx = -1
    min_rt = 999999

    for i in range(n):
        if at[i] <= current_time and rt[i] > 0:
            if rt[i] < min_rt:
                min_rt = rt[i]
                idx = i

    if idx != -1:
        rt[idx] = rt[idx] - 1
        current_time = current_time + 1

        if rt[idx] == 0:
            completed = completed + 1
            ct[idx] = current_time
            tat[idx] = ct[idx] - at[idx]
            wt[idx] = tat[idx] - bt[idx]
    else:
        current_time = current_time + 1

print("\nPID\tAT\tBT\tCT\tTAT\tWT")
for i in range(n):
    print(f"P{i+1}\t{at[i]}\t{bt[i]}\t{ct[i]}\t{tat[i]}\t{wt[i]}")

print("\nAVG TAT =", sum(tat) / n)
print("AVG WT =", sum(wt) / n)



```
